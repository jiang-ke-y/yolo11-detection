# YOLO11 Object Detection — Quick Start on Colab
Train and run inference with YOLO11 on a free GPU. Just **Run All**.

```python
# 1) Install Ultralytics (includes torch, opencv, etc.)
%pip install -q ultralytics
```

```python
# 2) Select a GPU at runtime: Runtime > Change runtime type > GPU
import torch
print('CUDA available:', torch.cuda.is_available())
```

```python
# 3) Train on the small COCO8 sample dataset (auto-downloaded)
from ultralytics import YOLO

model = YOLO('yolo11n.pt')
model.train(data='coco8.yaml', epochs=30, imgsz=640, device=0)
```

```python
# 4) Validate and print mAP metrics
metrics = model.val(data='coco8.yaml')
print('mAP50-95:', metrics.box.map)
```

```python
# 5) Run inference on a sample image and show the result
from google.colab.patches import cv2_imshow
import urllib.request

url = 'https://ultralytics.com/images/bus.jpg'
urllib.request.urlretrieve(url, 'bus.jpg')

results = model.predict('bus.jpg', conf=0.25, verbose=False)
cv2_imshow(results[0].plot())
```

```python
# 6) (Optional) export the trained model to ONNX
path = model.export(format='onnx', dynamic=True)
print('exported:', path)
```
